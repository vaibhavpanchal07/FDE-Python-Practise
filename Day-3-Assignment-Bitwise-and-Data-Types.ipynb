## 1. Assignment Operators

Assignment operators are used to assign values to variables. Augmented assignment operators combine an operation with assignment.

### Addition assignment (`+=`)

`x += 5` is equivalent to `x = x + 5`.

````python
x = 10
x += 5
print(x)
````

**Output:** `15`

### Subtraction assignment (`-=`)

`x -= 5` is equivalent to `x = x - 5`.

````python
x = 20
x -= 5
print(x)
````

**Output:** `15`

### Multiplication assignment (`*=`)

`x *= 3` is equivalent to `x = x * 3`.

````python
x = 10
x *= 3
print(x)
````

**Output:** `30`

### Division assignment (`/=`)

`x /= 5` is equivalent to `x = x / 5`. The `/` operator returns a floating-point result.

````python
x = 17
x /= 5
print(x)
````

**Output:** `3.4`

### Floor division assignment (`//=`)

`x //= 5` is equivalent to `x = x // 5`.

````python
x = 17
x //= 5
print(x)
````

**Output:** `3`

### Modulo assignment (`%=`)

`x %= 6` is equivalent to `x = x % 6`. It stores the remainder.

````python
x = 10
x %= 6
print(x)
````

**Output:** `4`

### Exponentiation assignment (`**=`)

`x **= 2` is equivalent to `x = x ** 2`.

````python
x = 5
x **= 2
print(x)
````

**Output:** `25`

### Assignment operators — quick reference

| Operator | Equivalent expression |
|---|---|
| `x += y` | `x = x + y` |
| `x -= y` | `x = x - y` |
| `x *= y` | `x = x * y` |
| `x /= y` | `x = x / y` |
| `x //= y` | `x = x // y` |
| `x %= y` | `x = x % y` |
| `x **= y` | `x = x ** y` |

## 3. Left Shift and Right Shift

Shift operators move the bits of an integer to the left or right.

### Left shift (`<<`)

The left shift operator moves bits to the left. Zeroes are added on the right.

Example: `15` in binary is `1111`.

```text
Before:  00001111   (15)
Shift:   << 1
After:   00011110   (30)
```

Each left shift by one position multiplies a non-negative integer by 2.

````python
# Left shift
x = 15
x <<= 1
print(x)
````

**Output:** `30`

### Another left shift example

`19` in binary is `10011`. Shifting left by two positions gives `1001100` (76).

````python
x = 19  # Binary: 10011
x <<= 2
print(x)
````

**Output:** `76`

### Try it yourself

````python
x = 21
x <<= 3
print(x)
````

**Expected output:** `168`

### Left shift reference

```text
15       = 1111
15 << 1  = 11110   (30)
15 << 2  = 111100  (60)
15 << 3  = 1111000 (120)
```

### Right shift (`>>`)

The right shift operator moves bits to the right. Bits shifted out on the right are discarded.

Example: `15` in binary is `1111`.

```text
Before:  1111  (15)
Shift:   >> 1
After:   0111  (7)
```

````python
# Right shift
x = 15
x >>= 1
print(x)
````

**Output:** `7`

### Right shift by four positions

````python
x = 15
x >>= 4
print(x)
````

**Output:** `0`

The original binary representation `1111` has only four bits. Shifting it right by four positions leaves no set bits.

### Right shift reference

```text
15       = 1111
15 >> 1  = 0111  (7)
15 >> 2  = 0011  (3)
15 >> 3  = 0001  (1)
15 >> 4  = 0000  (0)
```

**Remember:** A left shift adds zero bits on the right. A right shift discards bits from the right.

## 4. Bitwise Operators

Bitwise operators perform operations on the individual binary bits of integers.

The main bitwise operators are:

- Bitwise AND (`&`)
- Bitwise OR (`|`)
- Bitwise XOR (`^`)
- Bitwise NOT (`~`)

### Bitwise AND (`&`)

The result bit is `1` only when both corresponding bits are `1`.

Example:

````python
a = 5
b = 3

print("Bitwise AND:", a & b)
````

**Output:** `Bitwise AND: 1`

Binary breakdown:

```text
  5 = 0101
  3 = 0011
    -------
AND = 0001  -> 1
```

### Task

Try the same operation with `a = 8` and `b = 4`.

````python
a = 8
b = 4

print("Bitwise AND:", a & b)
````

**Output:** `Bitwise AND: 0`

```text
  8 = 1000
  4 = 0100
    -------
AND = 0000  -> 0
```

### Bitwise OR (`|`)

The result bit is `1` when at least one of the corresponding bits is `1`.

````python
a = 5
b = 3

print("Bitwise OR:", a | b)
````

**Output:** `Bitwise OR: 7`

```text
  5 = 0101
  3 = 0011
    -------
 OR = 0111  -> 7
```

### Bitwise XOR (`^`)

The result bit is `1` when the corresponding bits are different. If both bits are the same, the result is `0`.

````python
a = 5
b = 3

print("Bitwise XOR:", a ^ b)
````

**Output:** `Bitwise XOR: 6`

```text
  5 = 0101
  3 = 0011
    -------
XOR = 0110  -> 6
```

### Bitwise NOT (`~`)

The bitwise NOT operator reverses the bits. In Python, for an integer `a`, the result is `-(a + 1)`.

Example:

````python
a = 6
print("Bitwise NOT:", ~a)
````

**Output:** `Bitwise NOT: -7`

Because:

```text
~a = -(a + 1)
~6 = -(6 + 1)
   = -7
```

# Day 3 — Data Types in Python

A data type tells us what kind of value a variable refers to. Python provides several built-in data types.

The instructor's list:

1. `int` — integer
2. `float`
3. `complex`
4. `bool`
5. `str`
6. `list`
7. `tuple`
8. `set`
9. `dict`
10. `range`

We can use the `type()` function to check the type of a value.

## 1. Integer (`int`)

An integer represents a whole number. It can be positive, negative, or zero.

````python
age = 25
count = 200
temp = -10

print(type(temp))
````

**Output:** `<class 'int'>`

## 2. Float (`float`)

A float represents a number with a decimal point.

````python
salary = 10.0
print(type(salary))
````

**Output:** `<class 'float'>`

## 3. Complex (`complex`)

A complex number has a real part and an imaginary part. Python uses `j` for the imaginary component.

````python
number = 2 + 3j
print(type(number))
print(number.real)
print(number.imag)
````

**Output**
```text
<class 'complex'>
2.0
3.0
```

Here, `2` is the real component and `3j` is the imaginary component.

## 4. Boolean (`bool`)

A Boolean value represents one of two values: `True` or `False`.

````python
is_active = True
is_completed = False

print(type(is_active))
````

**Output:** `<class 'bool'>`

## 5. String (`str`)

A string represents text. Strings can be written using single quotes, double quotes, or triple quotes.

````python
name = "veera"
print(type(name))
````

**Output:** `<class 'str'>`

## 6. List (`list`)

A list stores multiple items in a single variable. Lists are ordered and mutable, which means their contents can be changed.

````python
number = [10, 20, 30]
print(type(number))
print(number)
````

**Output**
```text
<class 'list'>
[10, 20, 30]
```

## 7. Tuple (`tuple`)

A tuple stores multiple items. Tuples are ordered and immutable, which means their contents cannot be changed after creation.

````python
coordinates = (10, 20, 30)
print(type(coordinates))
print(coordinates)
````

**Output**
```text
<class 'tuple'>
(10, 20, 30)
```

## 8. Set (`set`)

A set stores unique items. It does not preserve items in a guaranteed positional order, and duplicate values are removed.

````python
numbers = {10, 20, 20, 30}
print(type(numbers))
print(numbers)
````

**Output**
```text
<class 'set'>
{10, 20, 30}
```

The order in which set elements are displayed may vary.

## 9. Dictionary (`dict`)

A dictionary stores data in key-value pairs. Keys are used to access their corresponding values.

````python
user = {
    "name": "veera",
    "age": 23
}

print(type(user))
print(user)
print(user["name"])
````

**Output**
```text
<class 'dict'>
{'name': 'veera', 'age': 23}
veera
```

## 10. Range (`range`)

A range represents a sequence of integers. It is commonly used with loops.

````python
numbers = range(1, 5)

print(type(numbers))
print(list(numbers))
````

**Output**
```text
<class 'range'>
[1, 2, 3, 4]
```

`range(1, 5)` starts at `1` and stops before `5`.
