# Day 2 — Literals, Constants and Operators

## 1. Literals

A **literal** is a value written directly in the code.

Examples of common Python literals:

````python
# Integer literal
age = 15

# Floating-point literal
price = 99.99

# String literal
name = "FDE"

# Boolean literal
is_active = True

# None literal
result = None

# Collection literals
number = [1, 2, 3]                    # List: ordered and mutable
coordinate = (10, 20)                 # Tuple: ordered and immutable
unique_value = {1, 2, 3}              # Set: stores unique elements
user = {"name": "veera", "age": 23}   # Dictionary: key-value pairs

print(age, price, name, is_active, result)
print(number)
print(coordinate)
print(unique_value)
print(user)
````

Remember:
- A list is ordered and mutable.
- A tuple is ordered and immutable.
- A set stores unique elements.
- A dictionary stores key-value pairs.

## 2. Constants

A **constant** is a value that we intend not to change during program execution.

Python does not enforce constants through a special declaration keyword. By convention, we write constant names in uppercase letters, using underscores between words.

**Example**

````python
age = 25  # Variable

# Constants by naming convention
DB_PORT = 3306
PI = 3.14

print(age)
print(DB_PORT)
print(PI)
````

`DB_PORT` and `PI` are written in uppercase to communicate that they are intended to remain unchanged. This is a convention; Python still allows reassignment.

## 3. Operators

Operators are special symbols or keywords used to perform operations on values and variables.

### Types of operators

- **Arithmetic operators:** `+`, `-`, `*`, `/`, `//`, `%`, `**`
- **Assignment operators:** `=`, `+=`, `-=`, `*=`, `/=`
- **Comparison operators:** `==`, `!=`, `>`, `<`, `>=`, `<=`
- **Logical operators:** `and`, `or`, `not`
- **Bitwise operators:** `&`, `|`, `~`, `<<`, `>>`
- **Identity operators:** `is`, `is not`

## 4. Arithmetic operators

Arithmetic operators are used to perform mathematical operations.

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor division |
| `%` | Modulo (remainder) |
| `**` | Power |

### Addition

````python
a = 3
b = 4
c = a + b
print(c)
````

**Output:** `7`

### Subtraction

````python
a = 5
b = 7
d = a - b
print(d)
````

**Output:** `-2`

### Multiplication

````python
a = 4
b = 3
print(a * b)
````

**Output:** `12`

### Division (`/`)

The `/` operator performs division and returns a floating-point result.

````python
x = 21
y = 11
print(x / y)
````

**Output:** `1.9090909090909092`

### Floor division (`//`)

Floor division rounds the result down toward negative infinity.

````python
x = 21
y = 11
print(x // y)
````

**Output:** `1`

Notice what happens with a negative number:

````python
x = -9
y = 5

print(x / y)
print(x // y)
````

**Output**
```text
-1.8
-2
```

`-9 // 5` gives `-2`, not `-1`, because floor division rounds down toward negative infinity.

### Modulo (`%`)

The modulo operator returns the remainder after division.

````python
x = 4
y = 3
print(x % y)

x = 10
y = 3
print(x % y)
````

**Output**
```text
1
1
```

If the dividend is smaller than the divisor, the remainder is the dividend (for these positive values).

````python
x = 10
y = 20
print(x % y)

x = 20
y = 30
print(x % y)
````

**Output**
```text
10
20
```

### Exponentiation (`**`)

The `**` operator is used to calculate a power.

````python
print(2 ** 3)  # 2 x 2 x 2
print(3 ** 4)  # 3 x 3 x 3 x 3
````

**Output**
```text
8
81
```

## 5. Comparison operators

Comparison operators compare two values and return a Boolean result: `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

### Equal to (`==`)

````python
a = 5
b = 10
print(a == b)
````

**Output:** `False`

### Not equal to (`!=`)

````python
a = 5
b = 10
print(a != b)
````

**Output:** `True`

### Greater than (`>`)

````python
a = 10
b = 10
print(a > b)
````

**Output:** `False`

Both values are equal. Therefore, `a` is not greater than `b`. Use `>=` when equality should also be accepted.

````python
a = 10
b = 10
print(a >= b)
````

**Output:** `True`

### Less than (`<`)

````python
a = 20
b = 10
print(a < b)
````

**Output:** `False`

### Less than or equal to (`<=`)

````python
a = 40
b = 40
print(a <= b)
````

**Output:** `True`

### Example: Voting eligibility

A comparison can be used in a condition.

````python
age = 18

if age >= 18:
    print("you are eligible to vote.")
else:
    print("you are not eligible to vote.")
````

**Output**
```text
you are eligible to vote.
```

## 6. Logical operators

Logical operators are used to combine or reverse conditions.

### `and` operator

Remember: **both conditions must be true** for the complete `and` expression to be true.

| Condition 1 | Condition 2 | Result |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

**Examples**

````python
a = 10
b = 20
print(a > 5 and b > 15)  # True and True -> True

a = 5
b = 20
print(a >= 5 and b > 15) # True and True -> True

a = 5
b = 20
print(a > 5 and b > 15)  # False and True -> False
````

### `or` operator

Remember: **at least one condition must be true** for the complete `or` expression to be true.

| Condition 1 | Condition 2 | Result |
|---|---|---|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

**Examples**

````python
a = 3
b = 20
print(a > 5 or b > 15)  # False or True -> True

a = 3
b = 20
print(a > 5 or b < 15)  # False or False -> False
````

### `not` operator

The `not` operator reverses the Boolean result.

- `not True` becomes `False`.
- `not False` becomes `True`.

These are expressions, so write them as Python code:

````python
print(not True)
print(not False)

is_logged_in = False
print(not is_logged_in)
````

**Output**
```text
False
True
True
```

## 7. Assignment operators

Assignment operators assign a value to a variable. The `=` operator assigns a value; augmented assignment operators perform an operation and assign the result back.

| Operator | Example | Equivalent form |
|---|---|---|
| `=` | `a = 10` | Assign 10 to `a` |
| `+=` | `a += 5` | `a = a + 5` |
| `-=` | `a -= 5` | `a = a - 5` |
| `*=` | `a *= 5` | `a = a * 5` |
| `/=` | `a /= 5` | `a = a / 5` |

````python
a = 10
print(a)

a += 5
print(a)

a -= 3
print(a)

a *= 2
print(a)

a /= 4
print(a)
````

## 8. Bitwise operators

Bitwise operators perform operations on the binary representation of integers.

| Operator | Name |
|---|---|
| `&` | AND |
| `|` | OR |
| `~` | NOT |
| `<<` | Left shift |
| `>>` | Right shift |

A simple example:

````python
a = 5   # Binary: 0101
b = 3   # Binary: 0011

print(a & b)
print(a | b)
print(a << 1)
print(a >> 1)
print(~a)
````

For `5` (`0101`) and `3` (`0011`):
- `5 & 3` gives `1` (`0001`).
- `5 | 3` gives `7` (`0111`).
- `5 << 1` shifts the bits left and gives `10`.
- `5 >> 1` shifts the bits right and gives `2`.
- `~5` gives `-6` in Python's integer model.

## 9. Identity operators

Identity operators check whether two references point to the same object.

- `is` returns `True` when both names refer to the same object.
- `is not` returns `True` when they refer to different objects.

Example:

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)      # True: b refers to the same list as a
print(a is c)      # False: c is a different list object
print(a == c)      # True: the list values are equal
print(a is not c)  # True
````

**Important distinction:**
- `==` compares values.
- `is` compares object identity.

Use `==` for normal value comparisons. A common identity check is `value is None`.

## Quick revision

- A literal is a value written directly in code.
- Uppercase names are used by convention for constants.
- Arithmetic operators perform mathematical operations.
- Comparison operators return `True` or `False`.
- `and` requires both conditions to be true; `or` requires at least one; `not` reverses the result.
- Assignment operators assign or update variable values.
- Bitwise operators work on integer bits.
- Identity operators check whether references point to the same object.
