## 1. Identifiers

An **identifier** is a name used for a variable, function, class, module, or another program element.

Rules:
- May contain letters, digits and underscores.
- Cannot begin with a digit.
- Cannot contain spaces or punctuation such as `-` or `@`.
- Is case-sensitive (`name`, `Name`, and `NAME` differ).
- Cannot be a reserved keyword such as `if`, `class`, or `return`.

```python
customer_name = "Asha"
_private_value = 10
item2 = "Notebook"
MAX_RETRIES = 3
print(customer_name, _private_value, item2, MAX_RETRIES)
```

The following are intentionally invalid examples. Uncomment one line at a time to observe its `SyntaxError`.

````python
# 2items = "invalid"       # Cannot start with a digit
# customer-name = "Asha"   # Hyphen is not allowed in an identifier
# customer name = "Asha"   # Spaces are not allowed
# if = 3                   # 'if' is a keyword
````

### Naming conventions (PEP 8)

| Purpose | Convention | Example |
|---|---|---|
| Variable / function | `snake_case` | `customer_name`, `calculate_total()` |
| Class | `PascalCase` | `CustomerAccount` |
| Constant by convention | `UPPER_SNAKE_CASE` | `MAX_RETRIES` |
| Internal-use name | Leading underscore | `_cache` |

Conventions improve readability; they are not all enforced by Python. Uppercase names are **not** immutable constants.

## 2. Keywords versus built-in names

**Keywords** are part of Python syntax (`if`, `for`, `def`, `class`, `return`, `True`, `False`, `None`). **Built-ins** are names supplied by Python (`print`, `len`, `sum`, `list`, `str`). A built-in name is not necessarily a keyword, but overwriting it can break later code.

````python
import keyword

print(keyword.iskeyword("if"))     # True
print(keyword.iskeyword("print"))  # False
print(len([10, 20, 30]))           # 3

# Avoid shadowing built-ins:
# list = [1, 2, 3]
# list("abc")  # Would fail because 'list' now refers to a list object
letters = list("abc")
print(letters)
````

## 3. Variables and assignment

A useful mental model is that **a variable name is a label bound to an object**. Assignment evaluates the right-hand side and binds the name to the resulting object.

Python is dynamically typed: a name can later refer to an object of a different type.

````python
value = 42
print(value, type(value))

value = "forty-two"
print(value, type(value))
````

### Reassignment and aliases

Assigning one name to another does not necessarily copy the object. Both names may refer to the same object.

````python
first = [1, 2, 3]
second = first
second.append(4)

print("first:", first)
print("second:", second)
print("Same object?", first is second)
````

## 4. `type()`, `id()`, equality and identity

- `type(obj)` reports an object's type.
- `id(obj)` returns an identity value unique during that object's lifetime. In CPython it is commonly related to a memory address, but don't rely on that implementation detail.
- `==` compares values.
- `is` checks whether two references point to the same object.
- `is None` is the recommended way to test for `None`.

````python
a = [1, 2]
b = [1, 2]
c = a

print(a == b)  # True: equal contents
print(a is b)  # False: different list objects
print(a is c)  # True: same object
print(type(a), id(a), id(b))
````

## 5. Mutable versus immutable objects

**Mutable** objects can be changed in place. Common examples: `list`, `dict`, `set`.

**Immutable** objects cannot be changed in place. Common examples: `int`, `float`, `bool`, `str`, and `tuple` (the tuple's structure is immutable).

This distinction matters when objects are shared between names or passed to functions.

````python
# Strings are immutable; this creates a new string value.
name = "python"
# name[0] = "P"  # TypeError
name = "P" + name[1:]
print(name)

# Lists are mutable.
tasks = ["learn Python", "practice"]
tasks.append("write tests")
print(tasks)

original = {"status": "new"}
alias = original
alias["status"] = "done"
print(original)  # The shared dictionary was changed.
````

## 6. How Python executes code

Python is often described as interpreted. In the standard CPython implementation, source code is compiled to bytecode and then executed by the Python virtual machine. Other implementations can differ.

For practical debugging, distinguish:
- **Syntax error:** code cannot be parsed, e.g. a missing colon.
- **Runtime error:** execution begins but fails, e.g. division by zero.
- **Logical error:** code runs but produces the wrong result.

## 7. Inspecting values and production naming

`repr(value)` provides a developer-oriented representation; `isinstance(value, SomeType)` checks whether a value is an instance of a type.

In production code, prefer names that describe intent. For example, `request_timeout_seconds` is clearer than `x`; avoid names such as `list`, `str`, `sum`, or `id` because they shadow useful built-ins.

````python
request_timeout_seconds = 10
is_retryable = True
MAX_RETRY_ATTEMPTS = 3

print(repr("hello\nworld"))
print(isinstance(request_timeout_seconds, int))
````

## 8. Common mistakes checklist
- Using a keyword as a name or starting a name with a digit.
- Assuming names are case-insensitive.
- Assuming uppercase names are enforced constants.
- Assuming assignment always copies an object.
- Using `is` instead of `==` for value comparison.
- Shadowing built-in names such as `list` or `sum`.
- Trusting saved notebook output without rerunning the cell.
