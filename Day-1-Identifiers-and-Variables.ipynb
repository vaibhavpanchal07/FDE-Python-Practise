## Python for FDE

Python is a programming language. A programming language allows humans to write instructions that a computer can execute.

**What is programming?** Programming is the process of writing a set of instructions to communicate with a computer and make it perform a task.

### 1. Python is a high-level language

A high-level programming language is easier for humans to read, write, and understand.

Python uses English-like instructions, so we do not need to write complex machine-level instructions directly.

**Example**

````python
print("Hello World")
````

**Output**
```text
Hello World
```

Python makes programming easier for humans.

**Remember:** High-level means human-friendly.

### 2. Python is dynamically typed

Dynamic typing means that we do not need to declare a variable's data type explicitly. Python determines the type from the value assigned to it.

**Example**

````python
a = 10
print(a)
print(type(a))
````

In a language such as C, we normally declare the type:

```c
int a = 10;
```

In Python, we can simply write `a = 10`.

### 3. Python is commonly described as an interpreted language

We call Python an interpreted language because Python code is executed through the Python interpreter. We do not normally compile and run programs manually as we might in C or Java.

The Python interpreter handles the compilation and execution process for us.

### Complete program execution flow

A simplified view of the standard CPython execution flow:

```text
Python Source Code (.py)
          |
          v
   Python Compiler
   Source -> Bytecode
          |
          v
 Python Virtual Machine
   Executes Bytecode
          |
          v
     Program Output
```

The important point is that Python source code is compiled to bytecode, and the Python Virtual Machine executes that bytecode. So, “interpreted” does not mean that Python skips compilation completely.

### Before writing code: create an algorithm

An **algorithm** is a step-by-step procedure for solving a problem or completing a task.

Before writing a program, understand the problem and decide the steps needed to solve it. Then translate those steps into Python code.

**Example: Add two numbers**

**Step 1: Understand the problem**

We need to add two numbers and display the result.

**Step 2: Write the algorithm**

1. Start.
2. Get the first number.
3. Get the second number.
4. Add the two numbers.
5. Display the result.
6. Stop.

**Step 3: Convert the algorithm into Python code**

````python
num1 = 5
num2 = 10

result = num1 + num2

print("Sum:", result)
````

**Output**
```text
Sum: 15
```

**Another example: Prepare Maggi**

Algorithm (step-by-step instructions):

4. Turn on the stove and boil the water.

# Day 1 — Identifier and Variable

## 1. Identifier

An **identifier** is a name used to identify a program element, such as a variable, function, or class.

## 2. Variable

A **variable** is a name that refers to an object in memory during program execution.

Example:

````python
name = "Python"
age = 25

print(name)
print(age)
````

Here, `name` and `age` are identifiers used as variable names.

## Rules for defining a variable or identifier

- Use letters, digits, and underscores (`_`).
- Never start an identifier with a digit.
- Do not use Python keywords as identifiers, for example `if`, `for`, and `while`.
- Do not use spaces or hyphens (`-`) in an identifier.
- Prefer `snake_case` for variable names, for example `server_name`.
- Avoid using Python built-in function names as variable names.

### Valid and invalid identifier examples

````python
# Valid identifiers
server_name = "api-server"
user_age = 25
_private_value = 10
item2 = "Notebook"

# Invalid identifiers (kept as comments so this cell can run)
# 2items = "Notebook"       # Cannot start with a digit
# user-name = "Asha"        # Hyphen is not allowed
# user name = "Asha"        # Spaces are not allowed
# if = 3                    # 'if' is a Python keyword
````

## 3. Keywords

**Keywords** are reserved words in Python. They have a special meaning in Python syntax and cannot be used as ordinary identifiers.

Examples: `if`, `else`, `for`, `while`, `class`, `def`, `return`, `True`, `False`, and `None`.

We can use the `keyword` module to see Python's keyword list.

````python
import keyword

a = keyword.kwlist
print(a)
print(len(a))
````

The number of keywords can differ between Python versions. The code above displays the keywords in the Python version you are running.

### Example: A keyword cannot be used as a variable name

The following is intentionally invalid Python code. It is shown to demonstrate the `SyntaxError`; do not run it as a normal working example.

````python
# if = 3
# print(if)
````

## 4. Built-ins

Python provides ready-to-use built-in functions, types, and other names. Examples include `print()`, `len()`, `type()`, `list()`, and `sum()`.

We can inspect the built-in names using the `builtins` module.

````python
import builtins

print(dir(builtins))
````

### Example: Avoid using a built-in name as a variable

Python may allow you to assign a value to a name such as `len`, but doing so hides the built-in function in the current scope.

**Example**

````python
len = 4
print(len)
````

The next example shows the problem. Because `len` now refers to the integer `4`, Python cannot call it like the original `len()` function.

This cell is intentionally commented out because running it after assigning `len = 4` would raise a `TypeError`.

````python
# print(len("FDE"))  # TypeError: 'int' object is not callable
````

This is not a hard rule that Python enforces for every built-in name, but it is a good practice to avoid conflicts with built-ins. Do not use built-in names as variable names.

### Check whether a name is a keyword or a built-in

````python
import keyword
import builtins

print(keyword.iskeyword("if"))    # True: 'if' is a keyword
print(keyword.iskeyword("len"))   # False: 'len' is not a keyword

print("print" in dir(builtins))    # True: 'print' is a built-in
print("if" in dir(builtins))       # False: 'if' is a keyword, not a built-in
````

## Important note

The same naming rules apply to variables and identifiers in Python because a variable name is an identifier.

However, remember that **not all identifiers are variable names**. Identifiers can also name functions, classes, and other programming elements.

### Example: Function identifier

````python
def my_function():
    a = 3
    b = 4
    c = a + b
    return c

print(my_function())
````

**Output**
```text
7
```

Here, `my_function` is an identifier used as a function name. The names `a`, `b`, and `c` are identifiers used as local variable names.
